```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = 'https://raw.githubusercontent.com/bhavanachitragar/Superstore-Data-Analysis-using-AWS/main/Sample%20-%20Superstore.csv'
```

```python
df = pd.read_csv(url, encoding='windows-1252')
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
```

```python
df.info()
```

```python
df['Order Year']=df['Order Date'].dt.year
df['Order Month']=df['Order Date'].dt.month
```

```python
df.head()
```

```python
df.columns
```

```python
sales_by_year=df.groupby('Order Year')['Sales'].sum()
```

```python
sales_by_year
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(x=sales_by_year.index, y=sales_by_year.values, marker='o')
plt.title('Total Sales by Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.show()
```

```python
sales_by_month=df.groupby('Order Month')['Sales'].sum()
sales_by_month
```

```python
top_products=df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)
```

```python
top_products
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.lineplot(x=sales_by_month.index, y=sales_by_month.values, marker='o', color='green')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.subplot(1,2,2)
sns.barplot(x=top_products.values, y=top_products.index, palette='viridis')
plt.title('Top 10 Profitable Products')
plt.tight_layout()
plt.show()
```
